filtering the data by applying the conditions.

```python
import pandas as pd
data={
    'name':['abhiyan','ram','shyam','john'],
    'age':[20,30,18,25],
    'grade':[4.0,3.6,3.72,3.99],
    'marks':[98,88,93,96]

}
df= pd.DataFrame(data)

print("display the data of those whose age is 20 plus:\n")

ag= df[df['age']>19]

print(ag)
```

```python
# df.loc[row_index, "column name"]= new_value

import pandas as pd
data={
    'name':['abhiyan','ram','shyam','john'],
    'age':[20,30,18,25],
    'grade':[4.0,3.6,3.72,3.99],
    'marks':[98,88,93,96]

}
df= pd.DataFrame(data)

df.loc[1,'marks']= 94
print(df)
```
