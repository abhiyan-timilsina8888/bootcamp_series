```python
import pandas as pd
data={
    "name":["abhiyan","aayusa","timilsina","dhikari","ronaldo"],
    "age":[22,45,43,22,43],
    "salary":[32323,65356,87320,47634,82673]

}

df= pd.DataFrame(data)

grouped= df.groupby("age")["salary"].sum()

print(grouped)
```
