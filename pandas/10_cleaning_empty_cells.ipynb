# find out the empty cells using ->.isnull()

```python
import pandas as pd
data={
    "name":['abhiyan',None,'sam','john','michael'],
    "age":[12,None,45,74,43],
    "salary":[325745,84648,2356.234,46343,46234],
    "reg_no.":[7380,None,2435,3264,8654]

}
df= pd.DataFrame(data)
print(df)

print("isnull returns boolean value:")
print(df.isnull())


print("isnull sum returns the sum")
print(df.isnull().sum())

```

# removing the empty cell using .dropna()

```python
import pandas as pd
import numpy as np

df= pd.read_csv("data.csv")
new_df = df.dropna()
print(df.isna().sum())

print(new_df)
print(new_df.isna().sum())
```
