# for single column

```python
import pandas as pd
data={
    "name":["kane","abhiyan","madric"],
    "age":[63,19,73],
    "salary":[19375,873824,82]
}

df= pd.DataFrame(data)
df.sort_values(by="age",ascending=False,inplace = True)
print("sorted value\n")
print(df)
```

# for multiple colummn

```python
import pandas as pd
data={
    "name":["kane","abhiyan","madric","ronaldo"],
    "age":[63,3,87,3],
    "salary":[19375,873824,82,9898]
}
df= pd.DataFrame(data)
df.sort_values(by=["age","salary"],ascending=[True,False],inplace= True)

print(df.to_string())
```
