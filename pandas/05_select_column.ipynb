# single column returns series while multiple column returns data frame.

```python
import pandas as pd
data={
    'name':['abhiyan','ram','shyam','john'],
    'age':[20,30,18,25],
    'grade':[4.0,3.6,3.72,3.99],
    'marks':[98,88,93,96]

}
df= pd.DataFrame(data)
print("Data frame: \n")
print(df)
print("\n Name column :")
name= df['name']
print(name)

```

# Multiple column returns Dataframe

```python
import pandas as pd
data={
    'name':['abhiyan','ram','shyam','john'],
    'age':[20,30,18,25],
    'grade':[4.0,3.6,3.72,3.99],
    'marks':[98,88,93,96]

}

df= pd.DataFrame(data)

subset= df[['name','grade']]
print(subset)
```

# adding the columns

```python
import pandas as pd
data={
    'name':['abhiyan','ram','shyam','john'],
    'age':[20,30,18,25],
    'grade':[4.0,3.6,3.72,3.99],
    'marks':[98,88,93,96]

}
df= pd.DataFrame(data,index=['a','b','c','d'])
df['salary']=[1000.0,2000199.019,74875,677823]
print(df)

```

# adding the columns using insert() method

```python
# insert(loc, column name, data )

import pandas as pd
data={
    'name':['abhiyan','ram','shyam','john'],
    'age':[20,30,18,25],
    'grade':[4.0,3.6,3.72,3.99],
    'marks':[98,88,93,96]

}
df= pd.DataFrame(data)
df.insert(2,'salary',[1000.0,2000199.019,74875,677823])
print(df)
```

# removing a column

```python
import pandas as pd
data={
    "name":['abhiyan','sam','john','michael'],
    "age":[12,45,74,43],
    "salary":[325745,2356.234,46343,46234],
    "reg_no.":[7380,2435,3264,8654]
}

df= pd.DataFrame(data)
print("before removing a column\n")

print(df)
print("\n")

print("after removing the column \n")

df.drop(columns=['reg_no.'],inplace= True) # inplace= True helps make changes in original data.
print(df)
```
