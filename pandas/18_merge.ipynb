```python
import pandas as pd

df_customer = pd.DataFrame({
    "customer_id":[1,2,3,5],
    "name":["abhiyan","ronaldo","aayusa","saman"]
})

df_order=pd.DataFrame({
    "customer_id":[1,2,4],
    "order":[8973,9283,7789]
})

df_merged= pd.merge(df_customer,df_order,on="customer_id",how="inner")
print(df_merged)
```

```python
import pandas as pd

df_customer = pd.DataFrame({
    "customer_id":[1,2,3,5],
    "name":["abhiyan","ronaldo","aayusa","saman"]
})

df_order=pd.DataFrame({
    "customer_id":[1,2,4],
    "order":[8973,9283,7789]
})

df_merged = pd.merge(df_customer,df_order,how="cross")
print(df_merged)
```
