Interpolate means estimating missing values between known data points.

```python
import pandas as pd
data={
    'time':[None,1,2,4,5],
    'value':[3,9,None,25,36]
}

df= pd.DataFrame(data)
print("before interpolation\n")
print(df)


print("after interpolation\n")
df['time']=df['time'].interpolate(method='linear')
df['value']=df['value'].interpolate(method='linear')

print(df)
```
