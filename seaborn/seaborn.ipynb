Loading a sample dataset with seaborn:

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df= sns.load_dataset("tips")
print(df.head())
```

Common Graphs:


a) scatter plot

```python
sns.scatterplot(data = df , x='total_bill',y='tip')
plt.show()
```

b) line plot

```python
sns.lineplot(data=df, x="size",y="tip")
plt.show()
```

c) Bar Plot

```python
sns.barplot(data=df, x="sex", y="tip")
plt.show()
```

d) Count Plot

```python
sns.countplot(data=df, x="day")
plt.show()
```

e) Histogram

```python
sns.histplot(data=df, x="total_bill")
plt.show()
```

f) Box Plot

It shows median, quartile, outliers.

```python
sns.boxplot(data=df, x='day', y='total_bill')
plt.show()
```

g) Violin Plot

```python
sns.violinplot(data=df,x='day', y='total_bill')
plt.show()
```

h) pair plot

```python
sns.pairplot(df)
plt.show()
```

h) Heat Map

annot = True displays the number too without it only colors are displayed.

```python
corr= df.corr(numeric_only = True)

sns.heatmap(corr, annot= True)
plt.show()
```

# Hue
It separates data by another category using different colors.

```python
sns.violinplot(data=df, x="total_bill",y="tip",hue="sex")
plt.show()
```

adding the style

```python
sns.scatterplot(data=df, x="total_bill",y="tip",style= "smoker")
plt.show()
```
