# ufunc is the univeral function based on ndarray objects.

- they are used to implement vectorization in numpy.

# vectorization  is the process of converting iterative statement into a vector based operation .



# without ufunc using python built-in zip() method.

```python
import numpy as np
a= [1,2,3,4,5]
b= [0,9,8,7,6]
c=[]

for i,j in zip(a,b):
    c.append(i + j)

print(c)


```

# with ufunc , we can use add( ) function

```python

import numpy as np

x= np.array([1,5,6,32])
y= np.array([23,5,2,7])

z= np.add(x,y)
print(z)




```
