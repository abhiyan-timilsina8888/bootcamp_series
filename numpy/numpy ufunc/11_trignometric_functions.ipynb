# numpy provides sin() cos() and tan() as trignometric functions for numpy ufunc.

```python
import numpy as np
x= np.sin(np.pi/2)
print(x)
```

```python
import  numpy as np
x= np.array([np.pi, np.pi/2,np.pi/3])
y= np.cos(x)
print(y)
```

```python
# convert radian into angle

import numpy as np
arr= np.array([np.pi,np.pi/2,np.pi/3])
res = np.rad2deg(arr)
print(res)
```

```python
#convert degree into radian

import numpy as np
arr= np.array([180,45,90,135])
res = np.deg2rad(arr)
print(res)
```

# finding the angles from value

```python
# arc....sin,cos,tan......()

import numpy as np
ar= np.array([1,2.5,1.5,0])
res= np.arctan(ar)
print(res)
```

# numpy provides hypot() function for hypotenous

```python
import numpy as np

base=3
perp= 4

x= np.hypot(base,perp)
print(x)
```
