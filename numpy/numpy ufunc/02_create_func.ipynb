# frompyfunc(). method is used to add python fucntion to numpy ufunc.
frompyfunc() takes the following argumnets:
function,input(number of inuput argumnet),output(number of output)

```python
import numpy as np

def m_add(x,y):
    return x+y

x= np.frompyfunc(m_add, 2 ,1)

print(x([1,2,3,4,5],[4,3,2,1,5]))
print(type(x))
```
