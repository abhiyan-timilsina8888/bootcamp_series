```python
import numpy as np
num1= 2
num2= 6

x= np.lcm(num1, num2)
print(x)
```

```python
# .reduce() is used for array or range of list

import numpy as np

arr= np.array([3,6,9,12])

x= np.lcm.reduce(arr)
print(x)
```

```python
# range 

import numpy as np

x= np.arange(1,11)

res = np.lcm.reduce(x)
print(res)
```
