A discrete difference means subtracting two successive element.

for [1,2,3,4] = diff() method works as [2-1, 3-2, 4-3] 

```python
import numpy as np
arr = np.array([1,2,3,4])
x=np.diff(arr)
print(x)
```

we can perform the same action multiple times by using n=..

```python
import numpy as np
arr = np.array([1,2,3,4])
x=np.diff(arr,n=2)
print(x)
```
