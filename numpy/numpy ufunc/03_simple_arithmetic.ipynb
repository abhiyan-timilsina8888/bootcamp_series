```python
# addition = np.add()

import numpy as np
arr1= np.array([1,2,3,4,5])
arr2= np.array([3,4,2,5,4])

new_arr = np.add(arr1,arr2)
print(new_arr)
```

```python
#subtraction = np.subtracte()

import numpy as np

arr1= np.array([1,2,3,4,5])
arr2= np.array([3,4,2,5,4])

new_arr = np.subtract(arr2,arr1)
print(new_arr)
```

```python
# multiplication = np.multiply()


import numpy as np

arr1= np.array([1,2,3,4,5])
arr2= np.array([3,4,2,5,4])

new_arr = np.multiply(arr2,arr1)
print(new_arr)
```

```python
# division = np.divide()

import numpy as np

arr1= np.array([1,2,3,4,5])
arr2= np.array([3,4,2,5,4])

new_arr = np.divide(arr2,arr1)
print(new_arr)
```

```python
#power = np.power()

import numpy as np

arr1= np.array([1,2,3,4,5])
arr2= np.array([3,4,2,5,4])

new_arr = np.power(arr2,arr1)
print(new_arr)
```

# mod or remainder does the same thing == np.mod() or np.remainder()

```python
import numpy as np

arr1= np.array([1,2,3,4,5])
arr2= np.array([3,4,2,5,4])

new_arr = np.mod(arr1,arr2)
new_arr1= np.remainder(arr1, arr2)
print(new_arr1)

print(new_arr)
```
