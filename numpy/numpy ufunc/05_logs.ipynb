# numpy takes three bases which are 2,10 and e 

```python
# for base 2 -> log2() is used 

import numpy as np
arr= np.arange(1,10)

x= np.log2(arr)
print(x)
```

```python
# for base 10 -> log10() is used

import numpy as np

x= np.log10(100)
print(x)
```

```python
# for natural log or base e log -> log() is used 

import numpy as np

x= np.arange(1,20)

arr= np.log(x)
print(arr)
```

for log with another base we don't have built in function 
so we use frompyfunc() along with inbuilt math.log() function


```python

import numpy as np
from math import log 

x= np.frompyfunc(log,2,1)

print(x(100,5)) # -> (value, base)



```
