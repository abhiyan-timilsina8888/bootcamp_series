prod() for product.

```python
import numpy as np
arr= np.array([1,2,3,4])
p= np.prod(arr)
print(p)
```

```python
import numpy as np
arr1= np.array([1,2,3,4])
arr2= np.array([5,6,7,8])
arr= np.prod([arr1,arr2])
print(arr)
```

```python
# product over an axis

import numpy as np
arr1= np.array([1,2,3,4])
arr2= np.array([5,6,7,8])
arr= np.prod([arr1,arr2],axis =1)
print(arr)
```

```python
#cummulative product

import numpy as np
arr1= np.array([1,2,3,4])
arr= np.cumprod(arr1)
print(arr)
```
