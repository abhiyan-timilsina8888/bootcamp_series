# five rounding methods 
1. truncation - removes the decimal part
2. fix - same as truncation. ---> [deprecated]
3. rounding - round off to nearest integer
4. ceil - round off to nearest upper integer
5. floor - rounf off to nearest lower integer


```python
# truncation and fix

import numpy as np
arr= np.trunc([-3.1666, 3.6667])
arr1= np.fix([-3.1666, 3.6667])
arr2= np.around([-3.1666,2.3])
arr3= np.floor([-3.1666, 3.6667])
arr4= np.ceil([-3.1666, 3.6667])
print(arr)
print(arr1)
print(arr2)
print(arr3)
print(arr4)



```
