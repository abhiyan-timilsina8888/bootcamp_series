indexing is used to obtain certain elemnt from the array.
[ ] is used for indexing.

starts from 0.
[1:4]= Here first elemnt is included while the last one is excluded.
[-1]= starts from the end.

```python
import numpy as np
arr=np.array([1,2,3,4])
print(arr[3])
```

```python
import numpy as np
arr= np.array([1,0,2,1,4])
print(arr[2]+arr[3])
```

```python
#negative indexing
import numpy as np 
arr=np.array([1,2,3,4])
print(arr[-2])
```
