Matplot library is a low level garph plotting library in python.
Is is a open source.

pyplot is a submodule of matplotlib , most of matplot utilities lies under pyplot.

```python
import numpy as np
import matplotlib.pyplot as plt
x= np.array([0,4])
y= np.array([0,100])

plt.plot(x,y)
plt.show()
```

By default plot() function draws line from point to point.

To plot the points without line we use  parameter 'o' which denotes as ring.

```python
import numpy as np
import matplotlib.pyplot as plt

x= np.array([1,4])
y=np.array([2,5])

plt.plot(x,y,'o')
plt.show()
```
