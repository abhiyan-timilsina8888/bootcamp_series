we can label the axis by uisng plt.x/y label

```python
import numpy as np
import matplotlib.pyplot as plt
x_arr=np.array([1,2,3,1,5])
y_arr=np.array([3,7,3,4,5])
```

```python
plt.plot(x_arr,y_arr)


plt.title("physics",loc='right')
plt.xlabel("distance")
plt.ylabel("velocity")

plt.show()
```

Set Font Properties for Title and Labels
fontdict parameter is used in xlabel(),ylabel() and title().

```python

font1={'family':'serif',
       'color':'darkred',
       'size':20}
font2={'family':'serif',
       'color':'black',
       'size':15}

plt.title("workout",fontdict=font1)
plt.xlabel("distance",fontdict=font2)
plt.ylabel("calories",fontdict=font2)

plt.plot(x_arr,y_arr)

plt.show()
```

plt.grid() is used for adding grids.
can use axis parameter to choose the axis for grid

```python
plt.grid(color='r',linewidth='0.4',ls=':')
plt.plot(x_arr,y_arr)
plt.show()
```
