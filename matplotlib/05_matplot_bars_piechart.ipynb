creating bars 

width for vetical bars

height for horizonatal bars.


```python
import numpy as np
import matplotlib.pyplot as plt

x= np.array(["A","B","C","D"])
y=np.array([5,3,7,3])

plt.bar(x,y,color='hotpink',width=0.6)
plt.show()
```

creating a horizontal bar

```python
import numpy as np
import matplotlib.pyplot as plt

x= np.array(["A","B","C","D"])
y=np.array([5,3,7,3])

plt.barh(x,y,color="#108414",height=0.4)

plt.show()
```

piechart

BY default it starts from 0 and continues counterclockwise.
we can specify the starting point using startangle.

Explode parameter helps make one of the wedge stand out.

Also shadow method is used to make it more realistic.

```python
import matplotlib.pyplot as plt
import numpy as np

y=np.array([25,35,40])
labels = np.array(["bmw","audi","ferrari"])
colors=np.array(["black","blue","hotpink"])
my_explode=np.array([0,0,0.2])
plt.pie(y,labels= labels,startangle=180,explode=my_explode,shadow=True,colors=colors)

plt.legend(title="cars",loc="upper right")
plt.show()
```
