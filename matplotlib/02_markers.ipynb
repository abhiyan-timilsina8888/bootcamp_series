```python
import numpy as np
import matplotlib.pyplot as plt
```

using marker for endpoints.

```python
y=np.array([2,1,9,5])

plt.plot(y, marker='*')
plt.plot()
```

other markers like
o for ring
D for diamond etc are also available.

formal string(fmt) is also used to specify the marker.
It is written in 
marker|line|color format.

here o denotes the marker
-. denotes line with dash and dot
y denotes the color yellow

```python
plt.plot(y, 'o-.y')
```

To manipulate the size of marker , 
-markersize or ms is used.
Also to color the markeredge,
-mec is used.
for marker face color
-mfc is used

```python
plt.plot(y,marker='D',ms=9,mec='k',mfc='hotpink')
plt.show()
```

linestyle(ls) is used for changing the style of plotted line.
color(c) is used for line.

```python
plt.plot(y,ls=':',c='hotpink',marker='o',mec='k',ms=4)
```

linewidth(lw) is used to amnipulate the width of line

```python
plt.plot(y,lw='10')
```
