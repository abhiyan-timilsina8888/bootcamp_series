using subplot() we can draw mutiple plot in same figure.

subplot() takes three argumnets;
1) row
2) column
3) index of 

we can create any numbers of plot just by specifying 
the no. of rows and columns.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Plot 1:

x=np.array([2,4,6,8])
y=np.array([1,3,5,7])

plt.subplot(1,2,1)
plt.title("income")
plt.plot(x,y)

#Plot 2:

x=np.array([1,3,5,7])
y=np.array([2,4,1,8])

plt.subplot(1,2,2)
plt.title("expenditure")
plt.plot(x,y)

# Adding the supertitle

plt.suptitle("plan")

plt.show()
```

# scatter plot

plt.scatter() is used to plot scatter points.

colors(c) argument is used to color them.

can compare multiple scatter plots as below:

```python
import numpy as np
import matplotlib.pyplot as plt

# day1: record of distance and calories

dis= np.array([9,4,8,2])
cal= np.array([100,40,80,15])

plt.scatter(dis,cal,c='green')

# day2: record of disatnce and calories

dis= np.array([5,14,8,12])
cal= np.array([50,140,8,115])

plt.scatter(dis,cal,c='darkred')

plt.show()

```

To color each dots

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

colors =np.array(
["red","green","blue","yellow","pink","black","orange","purple","beige","brown","gray","cyan","magenta"])

plt.scatter(x,y,c=colors)

plt.show()
```

colormap

It is like a list of colors that ranges from 0 to 100.

cmap argumnet is passed inside the scatter function for colormap.

THERE are different colorbars , one of them is viridis.
we can include this in the drawing using plt.colorbar().

size arugument is passed for manipulating the size of each point.

```python
import numpy as np
import matplotlib.pyplot as plt

x= np.array([1,2,6,8,3])
y=np.array([8,3,6,8,2])

color=np.array([30,40,27,58,78])
size=np.array([20,39,66,67,30])

plt.scatter(x,y,cmap='viridis',s=size,c=color)

plt.colorbar()

plt.show()
```

Alpha is used for adjusting the transparency of the dots using alpha argumnet.

let's combine color,size, and alpha.

```python
import numpy as np
from numpy import random
import matplotlib.pyplot as plt

x= np.random.randint(100,size=50)
y= np.random.randint(100,size=50)

color= np.random.randint(100,size=50)

size= 8 * np.random.randint(100,size=50)

plt.scatter(x,y,cmap='Set1',c=color, s=size, alpha=0.6)

plt.colorbar()
plt.show()

```
