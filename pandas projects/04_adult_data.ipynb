```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
df= pd.read_csv("adult.csv")
print(df.head())
print(df.shape)
```

```python
#5. Fetch Random Sample From the Dataset (50%)
data= df.sample(frac= 0.5,random_state= 42)
print(data)
```

```python
#6.Check Null Values In The Dataset
print(df.isnull().sum())
sns.heatmap(df.isnull())
plt.show()
```

```python
#7.Perform Data Cleaning [ Replace '?' with NaN ]
print(df.columns)
print(df.isin(['?']).sum())
df["workclass"]=df["workclass"].replace("?",np.nan)
df["occupation"]=df["occupation"].replace("?",np.nan)
df["native-country"]=df["native-country"].replace("?",np.nan)
sns.heatmap(df.isnull())
plt.show()
print(df.isin(["?"]).sum())

```

```python
#8. Drop all The Missing Values
print(df.isnull().sum())
print(df.dropna(how="any",inplace = True))
print(df.shape)
48842-45222
```

```python
#9. Check For Duplicate Data and Drop Them
print(df.duplicated().sum())

print(df.drop_duplicates(inplace= True))
print(df.duplicated().sum())
print("shape is:",df.shape)
```

```python
#10. Get Overall Statistics About The Dataframe
print(df.describe())
sns.displot(df.describe())
plt.show()
```

```python
# 11. Drop The Columns education-num, capital-gain, and capital-loss
print(df.columns)
df.drop(['educational-num','capital-gain','capital-loss'],axis=1,inplace= True)
print(df.columns)
```

```python
#12. What Is The Distribution of Age Column?
print(df["age"].describe())
print(df["age"].hist())
```

```python
#13. Find Total Number of Persons Having Age Between 17 To 48 (Inclusive) Using Between Method
print(df["age"].between(17,48).sum())
```

```python
#14. What is The Distribution of Workclass Column?
plt.figure(figsize=(10,10))
print(df["workclass"].hist())
```

```python
#15. How Many Persons Having Bachelors and Masters Degree?
print(df.columns)
# # print(df["education"].value_counts())
# data1= df["education"]=="Bachelors"
# print(data1.sum())
# data2= df["education"]=="Masters"
# print(data2.sum())
# data= data1| data2
# print(data.sum())

print(df["education"].isin(["Bachelors","Masters"]).sum())
```

```python
#16. Bivariate Analsis
plt.figure(figsize=(7,7))
sns.boxplot(x="income",y="age",data=df)
# print(df.info())
plt.show()
```

```python
#17. Replace income Values With 0 and 1
print(df["income"])
df["encoded_income"]=df['income'].replace(to_replace=["<=50K",">50K"],value=[0,1])
print(df.head(3))
```

```python
#18. Which Workclass Getting The Highest Salary?
print(df.columns)
# df["income"]=pd.to_numeric(df["income"],errors="coerce")
data= df.groupby("workclass")["encoded_income"].mean().sort_values(ascending= False)
print(data)
```

```python
# 19.How Has Better Chance To Get Salary greater than 50K Male or Female?
print(df.columns)
print(df["gender"].unique())
print(df.groupby("gender")["encoded_income"].mean().sort_values(ascending=False))
```

```python
#20. Covert workclass Columns Datatype To Category Datatype
df['workclass']=df['workclass'].astype('category')
print(df.info())
```
