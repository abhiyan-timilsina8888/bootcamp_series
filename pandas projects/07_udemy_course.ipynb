```python
import matplotlib.pyplot as plt
import seaborn as sns
print(df["subject"].value_counts())
sns.countplot(x="subject",data=df)
plt.xlabel("subject",fontsize=12)
plt.ylabel("count",fontsize=12)
plt.xticks(rotation=60)
plt.show()
```
