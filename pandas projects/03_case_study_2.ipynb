For data cleaning, errors='coerce' is the most commonly used option because it lets you convert as much data as possible while safely handling invalid entries.

```python
#12
df["BasePay"]=pd.to_numeric(df['BasePay'],errors ='coerce')
print(df["BasePay"].min())
print(df['BasePay'].max())
print(df['BasePay'].mean())
```

```python

#13  Replace 'Not Provided' in EmployeeName' Column to NaN 

# print(df.columns)
# print(df['EmployeeName'])
mask= df["EmployeeName"].str.contains("Not provided",case = False)
df.loc[mask,"EmployeeName"]="NaN"

print(df["EmployeeName"])
```

```python
#13 
import numpy as np
df["EmployeeName"]= df["EmployeeName"].replace("Not provided",np.nan)
df["EmployeeName"]= df["EmployeeName"].str.replace("Not provided","NaN",case=False,regex= False)
print(df.shape)


print(df["EmployeeName"])
```

```python
#14
print(df.shape)
print(df.isnull().sum(axis=1))
df.drop(df[df.isnull().sum(axis=1)==3].index,axis=0,inplace = True)
print(df)
```

```python
#15 Find Job Title of ALBERT PARDINI

# print(df[df["EmployeeName"].str.contains("ALBERT PARDINI",case= False)])
print(df[df["EmployeeName"]=="ALBERT PARDINI"]["JobTitle"])
```

```python
#16 How Much ALBERT PARDINI Make (Include Benefits)?
print(df.columns)
print(df[df["EmployeeName"].str.contains("ALBERT PARDINI",case= True)])
# print(df[df["EmployeeName"]=="ALBERT PARDINI"]["TotalPayBenefits"])
```

```python
# 17.Display Name of The Person Having The Highest BasePay
df["BasePay"]=pd.to_numeric(df["BasePay"],errors="coerce")
print(df[df["BasePay"]==df["BasePay"].max()]["EmployeeName"])
print(df[df["EmployeeName"].str.contains("Gregory P Suhr",case = True)])
```

```python
# 18.Find Average BasePay of All Employee Per Year 
print(df["BasePay"].mean())
```
