```python
print("how many people purchase during am and pm:\n")
d= df[df["AM or PM"]=="AM"]
e= df[df["AM or PM"]=="PM"]
print(len(d))
print(len(e))


'''usinf value_count() method to do this'''
print(df["AM or PM"].value_counts())
```

```python
print("how many credit cards were expireed in 2020:\n")

'''method first manually solving'''

def fun():
    count=0
    for data in df["CC Exp Date"]:
        if data.split('/')[1]=="20":
            count+=1
    print(count)

fun()
```

```python
'''method second using methods'''

print(len(df[df["CC Exp Date"].apply(lambda x: x[3:]=='20')]))
```

```python
print("top 5 most popular gmails:\n")
'''using function'''

def fun():
    list=[]
    for dat in df["Email"]:
        list.append(dat.split('@')[1])

    df['temp']=list
    print(df['temp'].value_counts().head())

fun()
```

```python

'''using lambda'''

print(df["Email"].apply(lambda x:x.split('@')[1]).value_counts().head())
```
