# 12. a) who Has Better Chance of Survival Male or Female?

```python
print(pd.crosstab(df["Sex"],df["Survived"]))
sns.countplot(x="Sex",hue="Survived",data= df)
```

```python
print(df[df["Survived"]==1]["Sex"].value_counts())
sns.countplot(x="Sex",hue="Survived",data=df)
plt.show()
```
