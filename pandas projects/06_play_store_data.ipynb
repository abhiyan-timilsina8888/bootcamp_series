```python
print(df.head())
```

# 2. Check the Last 3 Rows of The Dataset

```python
print(df.tail(3))
```

# 3. Find Shape of Our Dataset (Number of Rows & Number of Columns)

```python
print(df.shape)
print("rows:",df.shape[0])
print("columns:",df.shape[1])
```

# 4. Get Information About Our Dataset Like Total Number Rows, Total Number of Columns, Datatypes of Each Column And Memory Requirement

```python
print(df.info())
```

# 5. Get Overall Statistics About The Dataframe

```python
print(df.describe(include="all"))
```

# 6. Total Number of App Titles Contain Astrology

```python
df.columns
```

```python
# print(df["App"].value_counts())
print(sum(df["App"].str.contains("Astrology",case= False)))
```

# 7. Find Average App Rating

```python
print(df["Rating"].mean())
```

# 8. Find Total Number of Unique Category

```python
df.columns
```

```python
df["Category"].nunique()
```

# 9. Which Category Getting The Highest Average Rating?

```python
df.groupby("Category")["Rating"].mean().head().sort_values(ascending= False)
```

# 10. Find Total Number of App having 5 Star Rating

```python
print(df[df["Rating"]==5]["App"].value_counts().sum())
```

# 11. Find Average Value of Reviews

```python
df.columns
```

```python
df["Reviews"]=pd.to_numeric(df["Reviews"],errors="coerce")

print(df["Reviews"].mean())
```

# 12. Find Total Number of Free and Paid Apps

```python

df["Type"].value_counts()
```

# 13.  Which App Has Maximum Reviews?

```python
df.columns
print(df["Reviews"].max())
print(df[df["Reviews"]==df["Reviews"].max()]["App"])
```

# 14. Display Top 5 Apps Having Highest Reviews

```python
index=df["Reviews"].sort_values(ascending=False).head().index
df.iloc[index]["App"]
```

# 15. Find Average Rating of Free and Paid Apps

```python
df.columns
df["Type"].unique()
```

```python
print(df.groupby('Type').mean(numeric_only=True)["Rating"])
```

```python
print("rating of free app:",df[df["Type"]=="Free"].mean(numeric_only=True)["Rating"])
print("rating of paid app:",df[df["Type"]=="Paid"].mean(numeric_only=True)["Rating"])
```

# 16. Display Top  5 Apps Having Maximum Installs

```python
df.columns
print(df["Installs"].unique())
```

```python
df["Installs"]=df["Installs"].str.replace(r"[,+]","",regex= True)
df['Installs']=pd.to_numeric(df["Installs"],errors= "coerce")
print(df["Installs"].dtype)
```

```python
index=df["Installs"].sort_values(ascending= False).head().index
df.iloc[index]["App"]
```
